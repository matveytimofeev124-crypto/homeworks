```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
import matplotlib.pyplot as plt
```

```python
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return (x > 0).astype(float)

def softmax(x):
    exp_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return exp_x / np.sum(exp_x, axis=-1, keepdims=True)

def cross_entropy(y_pred, y_true, class_weights=None):
    log_pred = np.log(y_pred + 0.000000000001)
    if class_weights is not None:
        weights = class_weights.reshape(1, -1)
        loss = -np.mean(np.sum(weights * y_true * log_pred, axis=1))
        grad = weights * (y_pred - y_true)
    else:
        loss = -np.mean(np.sum(y_true * log_pred, axis=1))
        grad = y_pred - y_true
    return loss, grad
```

```python
class Layer:
    def __init__(self, n_inputs, n_neurons, activation='relu', l2=0.0):
        self.activation = activation
        self.l2_lambda = l2

        if activation == 'relu':
            self.W = np.random.randn(n_inputs, n_neurons) * np.sqrt(2. / n_inputs)
        else:
            self.W = np.random.randn(n_inputs, n_neurons) * 0.01
        self.b = np.zeros((1, n_neurons))

        self.X = None
        self.Z = None

        self.vW = np.zeros_like(self.W)
        self.vb = np.zeros_like(self.b)

    def forward(self, X):
        self.X = X
        self.Z = np.dot(X, self.W) + self.b

        if self.activation == 'relu':
            out = relu(self.Z)
        elif self.activation == 'softmax':
            out = softmax(self.Z)
        elif self.activation == 'linear':
            out = self.Z
        return out

    def backward(self, d_out, lr, momentum=0.9):
        if self.activation == 'relu':
            d_act = relu_derivative(self.Z)
        elif self.activation in ('linear', 'softmax'):
            d_act = 1.0

        dZ = d_out * d_act
        batch_size = self.X.shape[0]

        dW = np.dot(self.X.T, dZ) / batch_size
        db = np.mean(dZ, axis=0, keepdims=True)

        if self.l2_lambda > 0:
            dW += self.l2_lambda * self.W

        dX = np.dot(dZ, self.W.T)

        self.vW = momentum * self.vW - lr * dW
        self.vb = momentum * self.vb - lr * db
        self.W += self.vW
        self.b += self.vb

        return dX
```

```python
class NeuralNetwork:
    def __init__(self, layer_sizes, activations, l2=0.0):
        self.layers = []
        for i in range(len(layer_sizes) - 1):
            self.layers.append(Layer(layer_sizes[i], layer_sizes[i+1], activations[i], l2=l2))

    def forward(self, X):
        for layer in self.layers:
            X = layer.forward(X)
        return X

    def train(self, X_train, y_train, X_val=None, y_val=None,
              epochs=100, lr=0.01, batch_size=32, momentum=0.9,
              class_weights=None, verbose=True):
        n = X_train.shape[0]
        history = {'train_loss': [], 'val_acc': []}

        for epoch in range(epochs):
            idx = np.random.permutation(n)
            X_shuf, y_shuf = X_train[idx], y_train[idx]

            total_loss = 0.0
            num_batches = 0

            for i in range(0, n, batch_size):
                X_batch = X_shuf[i:i+batch_size]
                y_batch = y_shuf[i:i+batch_size]

                y_pred = self.forward(X_batch)

                loss, grad = cross_entropy(y_pred, y_batch, class_weights)
                total_loss += loss
                num_batches += 1

                for layer in reversed(self.layers):
                    grad = layer.backward(grad, lr, momentum)

            avg_loss = total_loss / num_batches
            history['train_loss'].append(avg_loss)

            if X_val is not None and y_val is not None:
                val_acc = self.accuracy(X_val, y_val)
                history['val_acc'].append(val_acc)
                if verbose:
                    print(f"Epoch {epoch+1}/{epochs}, loss: {avg_loss:.4f}, val_acc: {val_acc:.4f}")
            else:
                if verbose and (epoch+1) % 10 == 0:
                    print(f"Epoch {epoch+1}/{epochs}, loss: {avg_loss:.4f}")

        return history

    def predict_proba(self, X):
        return self.forward(X)

    def predict(self, X):
        return np.argmax(self.predict_proba(X), axis=1)

    def accuracy(self, X, y_true):
        preds = self.predict(X)
        true = np.argmax(y_true, axis=1)
        return np.mean(preds == true)
```

```python
data = pd.read_csv('WineQT.csv')

X = data.drop('quality', axis=1).values
y = data['quality'].values

scaler = StandardScaler()
X = scaler.fit_transform(X)

encoder = OneHotEncoder(sparse_output=False)
y_onehot = encoder.fit_transform(y.reshape(-1, 1))

X_train, X_test, y_train, y_test = train_test_split(X, y_onehot, test_size=0.2, random_state=42)

unique_classes = np.unique(y)
class_counts = np.array([np.sum(y == c) for c in unique_classes])
class_weights = 1.0 / class_counts
class_weights = class_weights / np.sum(class_weights) * len(unique_classes)

nn = NeuralNetwork(layer_sizes=[X.shape[1], 128, 64, y_onehot.shape[1]],
                    activations=['relu', 'relu', 'softmax'], l2=0.0001)

history = nn.train(X_train, y_train, X_val=X_test, y_val=y_test,
                    epochs=500, lr=0.01, batch_size=32, momentum=0.9,
                    class_weights=class_weights, verbose=True)
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
plt.plot(history['train_loss'], label='График потерь')
plt.xlabel('Эпоха')
plt.ylabel('Потери')
plt.title('Функция потерь')
plt.legend()
plt.grid()

plt.subplot(1, 2, 2)
plt.plot(history['val_acc'], label='График точности')
plt.xlabel('Эпоха')
plt.ylabel('Точность')
plt.title('Точность на валидации')
plt.legend()
plt.grid()

plt.tight_layout()
plt.show()

final_acc = nn.accuracy(X_test, y_test)
print(f"Финальная тестовая точность: {final_acc:.4f}")
```
