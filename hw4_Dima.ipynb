```python
# 1. Напишите функцию generate_password(length), которая генерирует случайный пароль заданной длины, содержащий:
# заглавные и строчные латинские буквы,
# цифры,
# символы !@#$%^&*.
# Функция должна гарантированно включать хотя бы один символ из каждой категории.

import random
import string

n = 12
def password(n: int) -> str:
    new_password = []
    alphabet = list(string.ascii_lowercase)
    alphabet1 = list(string.ascii_uppercase)
    digit = list(range(10))
    symbols = list('!@#$%^&*')
    password = []
    password.append(alphabet)
    password.append(alphabet1)
    password.append(digit)
    password.append(symbols)
    password1 = alphabet + alphabet1 + digit + symbols

    for i in range(4):
        new_password.append(password[i][random.randint(0, len(password[i]) - 1)])
    for i in range(n-4):
        new_password.append(random.choice(password1))
    random.shuffle(new_password)
    new_str = str(new_password)
    return new_str

password(n)
```

```python
# 2. Напишите программу, которая запрашивает у пользователя текст и выводит:
# количество слов, количество уникальных слов, самое часто встречающееся слово (и сколько раз), среднюю длину слова.

import string
from collections import Counter

text = input("Введите текст: ")
def text_count(text: str) -> str:
    word = str.maketrans('', '', string.punctuation)
    clean = text.translate(word).lower()
    words = clean.split()
    words_total = len(words)
    words_unique = len(set(words))
    words_count = Counter(words)
    words_common, count_common = words_count.most_common(1)[0]
    avr_length = sum(len(letter) for letter in words) / words_total
    print("Количество слов: ", words_total)
    print("Количество уникальных слов: ", words_unique)
    print("Самое часто встречающееся слово: ", words_common, "(встречается ", count_common, " раз)")
    print("Средняя длина слова: ", avr_length, " символов")

text_count(text)
```

```python
# 3. Компьютер загадывает число от 1 до 100. Пользователь пытается угадать. После каждой попытки компьютер говорит «больше», «меньше» или «угадал». В конце вывести количество попыток.

import random

n = random.randint(1, 100)
k = 0

while True:
    number = int(input())
    if number < n:
        print("Загаданное число больше!")
        k = k + 1
    elif number > n:
        print("Загаданное число меньше!")
        k = k + 1
    else:
        print("Число угадано!")
        k = k + 1
        break

print("Количество попыток: ", k)
```

```python
# 4. Напишите программу «Таймер». Пользователь вводит количество секунд. Программа отсчитывает время, каждую секунду выводя оставшееся время в формате MM:SS.
# По истечении времени выводится сообщение и проигрывается звуковой сигнал (print('\a')).

import time

def timer(seconds):
    while seconds > 0:
        minutes, secs = divmod(seconds, 60)
        print(f"{minutes:02d}:{secs:02d}", end="\r")
        time.sleep(1)
        seconds -= 1

    print("\nВремя вышло!")
    print('\a')

try:
    total_seconds = int(input("Введите количество секунд: "))
    if total_seconds < 0:
        print("Ошибка: количество секунд не может быть отрицательным.")
    else:
        timer(total_seconds)
except ValueError:
    print("Ошибка: введите целое число.")
```

```python
# 5. Напишите функцию calculate(expression), которая принимает строку вида "3 + 5 * 2" (поддерживаются +, -, *, /, ^ для степени)
# и возвращает результат с учётом приоритета операций. Используйте обратную польскую запись или рекурсивный спуск. eval использовать нельзя.

import re

def calculate(expression):
    prec = {'+': 1, '-': 1, '*': 2, '/': 2, '^': 3}
    right_assoc = {'^'}

    tokens = re.findall(r'\d+(?:\.\d+)?|[+\-*/^]', expression)

    output = []
    stack = []

    for tok in tokens:
        if tok.replace('.', '', 1).isdigit():
            output.append(tok)
        else:
            while (stack and stack[-1] != '(' and
                   (prec[stack[-1]] > prec[tok] or
                    (prec[stack[-1]] == prec[tok] and tok not in right_assoc))):
                output.append(stack.pop())
            stack.append(tok)

    while stack:
        output.append(stack.pop())

    vals = []
    for tok in output:
        if tok.replace('.', '', 1).isdigit():
            vals.append(float(tok))
        else:
            b = vals.pop()
            a = vals.pop()
            if tok == '+':    vals.append(a + b)
            elif tok == '-':  vals.append(a - b)
            elif tok == '*':  vals.append(a * b)
            elif tok == '/':  vals.append(a / b)
            elif tok == '^':  vals.append(a ** b)

    return vals[0]

calculate("2 * 3 ^ 2 + 4")
```
