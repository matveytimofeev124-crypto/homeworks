```python
# ============================================================
# 0. IMPORTS, CONFIG, LOGGING
# ============================================================
import warnings
warnings.filterwarnings("ignore")

import logging
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error

from statsmodels.tsa.stattools import adfuller, grangercausalitytests
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.api import VAR
from statsmodels.tsa.forecasting.stl import STLForecast
from statsmodels.tsa.arima.model import ARIMA

from prophet import Prophet
from tbats import TBATS

import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import LSTM, GRU, Dense, Dropout, Input
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

# --- Reproducibility ---
np.random.seed(42)
tf.random.set_seed(42)
logging.getLogger('prophet').setLevel(logging.WARNING)
logging.getLogger('cmdstanpy').setLevel(logging.WARNING)

sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (14, 5)

# --- Config ---
CONFIG = dict(
    TARGETS      = ['meantemp', 'humidity', 'wind_speed', 'meanpressure'],
    HORIZON      = 7,
    LOOKBACK     = 90,
    FOURIER_K    = 4,
    LAGS         = (1, 2, 3, 7, 14),
    TEST_LEN     = 114,
    TRAIN_PATH   = 'archive/DailyDelhiClimateTrain.csv',
    TEST_PATH    = 'archive/DailyDelhiClimateTest.csv',
    CLIP_BOUNDS  = {
        'meantemp':     (None, None),
        'humidity':     (0, 100),
        'wind_speed':   (0, None),
        'meanpressure': (950, 1050),
    },
)
TARGETS   = CONFIG['TARGETS']
HORIZON   = CONFIG['HORIZON']
LOOKBACK  = CONFIG['LOOKBACK']
FOURIER_K = CONFIG['FOURIER_K']
LAGS      = CONFIG['LAGS']
TEST_LEN  = CONFIG['TEST_LEN']
CLIP_BOUNDS = CONFIG['CLIP_BOUNDS']


# ============================================================
# 1. UTILITIES
# ============================================================
def evaluate(y_true, y_pred, name="model"):
    y_true = np.asarray(y_true).ravel()
    y_pred = np.asarray(y_pred).ravel()
    mae  = mean_absolute_error(y_true, y_pred)
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mape = np.mean(np.abs((y_true - y_pred) / (y_true + 1e-8))) * 100
    return {"Model": name, "MAE": mae, "RMSE": rmse, "MAPE": mape}


def calc_mape(a, p):
    a, p = np.asarray(a).ravel(), np.asarray(p).ravel()
    return float(np.mean(np.abs((a - p) / a)) * 100)


def fourier_terms(t_index, K=FOURIER_K, period=365.25):
    t = np.asarray(t_index, dtype=float)
    data = {}
    for k in range(1, K + 1):
        data[f'sin_{k}'] = np.sin(2 * np.pi * k * t / period)
        data[f'cos_{k}'] = np.cos(2 * np.pi * k * t / period)
    return pd.DataFrame(data)


def add_calendar(df):
    df = df.copy()
    doy = df.index.dayofyear.values
    df['doy_sin'] = np.sin(2 * np.pi * doy / 365.25)
    df['doy_cos'] = np.cos(2 * np.pi * doy / 365.25)
    return df


def get_d(series):
    """Автоподбор порядка дифференцирования по ADF."""
    if adfuller(series.dropna())[1] <= 0.05:
        return 0
    if adfuller(series.diff().dropna())[1] <= 0.05:
        return 1
    return 2


def clip_target(arr, target):
    lo, hi = CLIP_BOUNDS[target]
    if lo is None and hi is None:
        return arr
    return np.clip(arr, lo, hi)


def coverage(actual, lower, upper):
    inside = (actual >= lower) & (actual <= upper)
    return inside.mean() * 100


# ============================================================
# 2. DATA LOADING + CLEANING (включая фикс фейковых строк)
# ============================================================
def load_and_clean(path):
    df = pd.read_csv(path, parse_dates=['date']).set_index('date').asfreq('D')
    df.loc[(df['meanpressure'] < 950) | (df['meanpressure'] > 1100),
           'meanpressure'] = np.nan
    df['meanpressure'] = df['meanpressure'].interpolate(method='time').bfill().ffill()
    return df


def fix_fake_rows(train, test, targets):
    """
    В датасете DailyDelhiClimate последняя строка train (2017-01-01)
    заполнена placeholder-значениями [10.0, 100.0, 0.0, 1016.0],
    хотя реальные значения этого дня есть в test.
    """
    train = train.copy()
    fake_mask = (
        (train['meantemp'] == 10.0) &
        (train['humidity'] == 100.0) &
        (train['wind_speed'] == 0.0)
    )
    n_fake = int(fake_mask.sum())
    print(f"Найдено фейковых строк в train: {n_fake}")
    if n_fake == 0:
        return train

    for date in train.index[fake_mask]:
        if date in test.index:
            train.loc[date, targets] = test.loc[date, targets].values
            print(f"  {date.date()}: заменено значениями из test")
        else:
            train.loc[date, targets] = np.nan
            print(f"  {date.date()}: нет в test → интерполяция")

    train[targets] = train[targets].interpolate(method='time').bfill().ffill()
    return train


def load_data():
    train = load_and_clean(CONFIG['TRAIN_PATH'])
    test  = load_and_clean(CONFIG['TEST_PATH'])
    train = fix_fake_rows(train, test, TARGETS)

    assert np.allclose(
        train.loc['2017-01-01', TARGETS].values,
        test.loc['2017-01-01', TARGETS].values
    ), "Фикс фейковых строк не сработал!"
    print("✅ Фикс применён: train['2017-01-01'] == test['2017-01-01']")
    return train, test


# ============================================================
# 3. SPLIT (internal test — последние 114 дней train)
# ============================================================
def make_split(train):
    train_int = train.iloc[:-TEST_LEN]
    test_int  = train.iloc[-TEST_LEN:]
    print(f"Train: {train.shape}, Internal test: {test_int.shape}")
    return train_int, test_int


# ============================================================
# 4. MODEL: SARIMAX + Fourier
#    (walk-forward, финальный прогноз, bias-correction, CI — в одном блоке)
# ============================================================
class SARIMAXModel:
    def __init__(self, order=(1, 1, 1), seasonal_order=(1, 0, 1, 7)):
        self.order = order
        self.seasonal_order = seasonal_order
        self.bias = 0.0
        self.d_ = None

    # --- внутреннее обучение ---
    def _fit(self, series, exog):
        return SARIMAX(
            series.values, exog=exog.values,
            order=(self.order[0], self.d_, self.order[2]),
            seasonal_order=self.seasonal_order,
            enforce_stationarity=False, enforce_invertibility=False
        ).fit(disp=False)

    # --- walk-forward на 1 шаг (для внутреннего теста) ---
    def walk_forward(self, series_train, series_test,
                     exog_train=None, exog_test=None):
        history   = list(series_train.values)
        hist_exog = exog_train.reset_index(drop=True).copy()
        test_exog = exog_test.reset_index(drop=True).copy()
        preds = []
        for t in range(len(series_test)):
            try:
                m = SARIMAX(
                    history,
                    exog=hist_exog.iloc[:len(history)].values,
                    order=(self.order[0], self.d_, self.order[2]),
                    seasonal_order=self.seasonal_order,
                    enforce_stationarity=False, enforce_invertibility=False
                ).fit(disp=False)
                yhat = float(m.forecast(
                    steps=1, exog=test_exog.iloc[[t]].values).iloc[0])
            except Exception:
                yhat = history[-1]
            preds.append(yhat)
            history.append(series_test.values[t])
            hist_exog = pd.concat([hist_exog, test_exog.iloc[[t]]],
                                  ignore_index=True)
        return pd.Series(preds, index=series_test.index)

    # --- прямой прогноз на h шагов ---
    def forecast_h(self, series_train, horizon, exog_train=None, bias=None):
        m = self._fit(series_train, exog_train)
        t_future   = np.arange(len(series_train), len(series_train) + horizon)
        exog_fut   = fourier_terms(t_future)
        fc         = m.get_forecast(steps=horizon, exog=exog_fut.values)
        mean = np.asarray(fc.predicted_mean).ravel()
        ci   = np.asarray(fc.conf_int(alpha=0.05))
        b = self.bias if bias is None else bias
        return mean + b, ci + b

    # --- bias correction через backtest ---
    def compute_bias(self, series, n_recent=60):
        tr  = series.iloc[:-n_recent]
        val = series.iloc[-n_recent:]
        exog = fourier_terms(np.arange(len(tr)))
        m = self._fit(tr, exog)
        exog_f = fourier_terms(np.arange(len(tr), len(tr) + n_recent))
        preds = np.asarray(m.forecast(steps=n_recent,
                                      exog=exog_f.values)).ravel()
        self.bias = float(np.mean(val.values - preds))
        return self.bias

    # --- полный прогон по одной переменной ---
    def run_target(self, target, train_int, test_int, train_full):
        self.d_ = get_d(train_int[target])
        exog_tr = fourier_terms(np.arange(len(train_int)))
        exog_te = fourier_terms(np.arange(len(train_int),
                                           len(train_int) + len(test_int)))

        pred_int = self.walk_forward(train_int[target], test_int[target],
                                     exog_tr, exog_te)
        metrics  = evaluate(test_int[target].values, pred_int.values,
                            f"SARIMAX [{target}]")

        exog_full = fourier_terms(np.arange(len(train_full)))
        mean7, ci7 = self.forecast_h(train_full[target], HORIZON, exog_full)
        mean7 = clip_target(mean7, target)

        self.compute_bias(train_full[target], n_recent=60)
        mean7_bc = clip_target(mean7 + self.bias, target)

        return {
            "metrics":    metrics,
            "forecast7":  mean7,
            "forecast7_bc": mean7_bc,
            "ci7":        ci7,
            "bias":       self.bias,
            "pred_int":   pred_int,
        }


# ============================================================
# 5. MODEL: Prophet + lag-регрессоры
#    (recursive + direct, walk-forward, bias — в одном блоке)
# ============================================================
class ProphetModel:
    def __init__(self, lags=LAGS):
        self.lags = lags
        self.bias = 0.0

    def _prep_lags(self, df, lags):
        d = df.copy()
        for l in lags:
            d[f'lag_{l}'] = d['y'].shift(l)
        return d.dropna().reset_index(drop=True)

    def _fit(self, train_lagged):
        m = Prophet(daily_seasonality=False, weekly_seasonality=True,
                    yearly_seasonality=True, changepoint_prior_scale=0.01,
                    seasonality_prior_scale=0.1)
        for c in [f'lag_{l}' for l in self.lags]:
            m.add_regressor(c)
        m.fit(train_lagged)
        return m

    def _to_df(self, series_df, target):
        return (series_df.reset_index()[['date', target]]
                .rename(columns={'date': 'ds', target: 'y'}))

    # --- walk-forward (recursive) ---
    def walk_forward(self, train_df, test_df):
        lag_cols = [f'lag_{l}' for l in self.lags]
        history  = train_df[['ds', 'y']].copy()
        preds = []
        for i in range(len(test_df)):
            m = self._fit(self._prep_lags(history, self.lags))
            row = {'ds': test_df['ds'].iloc[i]}
            for l in self.lags:
                row[f'lag_{l}'] = history['y'].values[-l]
            preds.append(m.predict(pd.DataFrame([row]))['yhat'].iloc[0])
            history = pd.concat([history, test_df[['ds', 'y']].iloc[[i]]],
                                ignore_index=True)
        return pd.Series(preds, index=test_df['ds'].values)

    # --- recursive forecast ---
    def forecast_recursive(self, train_df, horizon):
        m = self._fit(self._prep_lags(train_df, self.lags))
        last_ds = train_df['ds'].iloc[-1]
        last_y  = list(train_df['y'].values)
        means, los, his = [], [], []
        for h in range(1, horizon + 1):
            row = {'ds': last_ds + pd.Timedelta(days=h)}
            for l in self.lags:
                row[f'lag_{l}'] = last_y[-l]
            pred = m.predict(pd.DataFrame([row]))
            means.append(pred['yhat'].iloc[0])
            los.append(pred['yhat_lower'].iloc[0])
            his.append(pred['yhat_upper'].iloc[0])
            last_y.append(pred['yhat'].iloc[0])
        return np.array(means), np.column_stack([los, his])

    # --- direct forecast (одна модель на каждый h) ---
    def forecast_direct(self, train_df, horizon):
        df = train_df.copy()
        predictions = {}
        for h in range(1, horizon + 1):
            df_h = df.iloc[::h].copy().reset_index(drop=True)
            for l in self.lags:
                df_h[f'lag_{l}'] = df_h['y'].shift(l)
            df_h = df_h.dropna().reset_index(drop=True)
            if len(df_h) < 30:
                predictions[h] = df['y'].iloc[-1]
                continue
            m = Prophet(daily_seasonality=False, weekly_seasonality=False,
                        yearly_seasonality=True, changepoint_prior_scale=0.01,
                        seasonality_prior_scale=0.1)
            for l in self.lags:
                m.add_regressor(f'lag_{l}')
            m.fit(df_h)
            row = {'ds': df['ds'].iloc[-1] + pd.Timedelta(days=h)}
            for l in self.lags:
                idx = -l * h
                row[f'lag_{l}'] = (df['y'].iloc[idx]
                                   if abs(idx) <= len(df) else df['y'].iloc[0])
            predictions[h] = m.predict(pd.DataFrame([row]))['yhat'].iloc[0]
        return np.array([predictions[h] for h in range(1, horizon + 1)])

    # --- bias ---
    def compute_bias(self, series, n_recent=60):
        tr  = series.iloc[:-n_recent]
        val = series.iloc[-n_recent:]
        means, _ = self.forecast_recursive(
            self._to_df(tr, 'meantemp'), n_recent)
        self.bias = float(np.mean(val.values - means))
        return self.bias

    # --- полный прогон по одной переменной ---
    def run_target(self, target, train_int, test_int, train_full):
        tr_int = self._to_df(train_int, target)
        te_int = self._to_df(test_int, target)
        pred_int = self.walk_forward(tr_int, te_int)
        metrics  = evaluate(test_int[target].values, pred_int.values,
                            f"Prophet [{target}]")

        tr_full = self._to_df(train_full, target)
        mean7, _ = self.forecast_recursive(tr_full, HORIZON)
        mean7 = clip_target(mean7, target)

        # bias только для meantemp (backtest на нём имеет смысл)
        if target == 'meantemp':
            self.compute_bias(train_full[target], n_recent=60)
        else:
            self.bias = 0.0
        mean7_bc = clip_target(mean7 + self.bias, target)

        return {
            "metrics":     metrics,
            "forecast7":   mean7,
            "forecast7_bc": mean7_bc,
            "bias":        self.bias,
            "pred_int":    pred_int,
        }


# ============================================================
# 6. MODEL: Multi-output LSTM / GRU
# ============================================================
class MultiRNNModel:
    NN_FEATURES = TARGETS + ['doy_sin', 'doy_cos']
    NN_TARGETS  = TARGETS

    def __init__(self, kind='lstm'):
        assert kind in ('lstm', 'gru')
        self.kind = kind
        self.model = None

    def _build(self):
        RNN = LSTM if self.kind == 'lstm' else GRU
        inp = Input(shape=(LOOKBACK, len(self.NN_FEATURES)))
        x = RNN(64, return_sequences=True)(inp)
        x = Dropout(0.3)(x)
        x = RNN(32)(x)
        x = Dropout(0.3)(x)
        x = Dense(32, activation='relu')(x)
        out = Dense(len(self.NN_TARGETS))(x)
        model = Model(inp, out)
        model.compile(optimizer=tf.keras.optimizers.Adam(1e-3),
                      loss='mse', metrics=['mae'])
        return model

    @staticmethod
    def _build_sequences(X, Y, lookback):
        Xs, Ys = [], []
        for i in range(len(X) - lookback):
            Xs.append(X[i:i + lookback])
            Ys.append(Y[i + lookback])
        return np.array(Xs), np.array(Ys)

    def _prepare(self, train_df, test_df, fit_scalers=True):
        tr_c, te_c = add_calendar(train_df), add_calendar(test_df)
        if fit_scalers or not hasattr(self, 'scaler_X'):
            self.scaler_X = MinMaxScaler().fit(tr_c[self.NN_FEATURES].values)
            self.scaler_Y = MinMaxScaler().fit(tr_c[self.NN_TARGETS].values)

        X_tr = self.scaler_X.transform(tr_c[self.NN_FEATURES].values)
        X_te = self.scaler_X.transform(te_c[self.NN_FEATURES].values)
        Y_tr = self.scaler_Y.transform(tr_c[self.NN_TARGETS].values)
        Y_te = self.scaler_Y.transform(te_c[self.NN_TARGETS].values)

        X_full = np.vstack([X_tr, X_te])
        Y_full = np.vstack([Y_tr, Y_te])
        X_all, Y_all = self._build_sequences(X_full, Y_full, LOOKBACK)
        split = len(X_tr) - LOOKBACK
        return (X_all[:split], Y_all[:split],
                X_all[split:], Y_all[split:])

    def fit(self, X_tr, Y_tr, X_va, Y_va):
        self.model = self._build()
        es = EarlyStopping(monitor='val_loss', patience=20,
                           restore_best_weights=True, verbose=0)
        rl = ReduceLROnPlateau(monitor='val_loss', factor=0.5,
                               patience=8, min_lr=1e-6, verbose=0)
        self.model.fit(X_tr, Y_tr, validation_data=(X_va, Y_va),
                       epochs=150, batch_size=32,
                       callbacks=[es, rl], verbose=0)
        return self

    def recursive_forecast(self, full_df, horizon=HORIZON):
        full_c = add_calendar(full_df)
        X_full = self.scaler_X.transform(full_c[self.NN_FEATURES].values)
        window = X_full[-LOOKBACK:].copy()
        last_date = full_df.index[-1]

        preds_all = []
        for h in range(1, horizon + 1):
            yhat_scaled = self.model.predict(
                window.reshape(1, LOOKBACK, -1), verbose=0)[0]
            preds_all.append(yhat_scaled)
            doy = (last_date + pd.Timedelta(days=h)).dayofyear
            new_step = np.concatenate([
                yhat_scaled,
                [np.sin(2 * np.pi * doy / 365.25),
                 np.cos(2 * np.pi * doy / 365.25)]
            ])
            window = np.vstack([window[1:], new_step])
        return self.scaler_Y.inverse_transform(np.array(preds_all))

    def run_target_all(self, train_int, test_int, train_full):
        """Обучаем на train_int → метрики на test_int; затем дообучаем на train."""
        # --- внутренний тест ---
        X_tr, Y_tr, X_te, Y_te = self._prepare(train_int, test_int)
        vs = int(len(X_tr) * 0.2)
        self.fit(X_tr[:-vs], Y_tr[:-vs], X_tr[-vs:], Y_tr[-vs:])
        pred_te  = self.scaler_Y.inverse_transform(
            self.model.predict(X_te, verbose=0))
        Y_te_orig = self.scaler_Y.inverse_transform(Y_te)

        metrics = {}
        for i, target in enumerate(TARGETS):
            metrics[target] = evaluate(
                Y_te_orig[:, i], pred_te[:, i], f"{self.kind.upper()} [{target}]")

        # --- финальное обучение на полном train ---
        tr_c = add_calendar(train_full)
        self.scaler_X = MinMaxScaler().fit(tr_c[self.NN_FEATURES].values)
        self.scaler_Y = MinMaxScaler().fit(tr_c[self.NN_TARGETS].values)
        X_full = self.scaler_X.transform(tr_c[self.NN_FEATURES].values)
        Y_full = self.scaler_Y.transform(tr_c[self.NN_TARGETS].values)
        X_all, Y_all = self._build_sequences(X_full, Y_full, LOOKBACK)
        vs = int(len(X_all) * 0.2)
        self.fit(X_all[:-vs], Y_all[:-vs], X_all[-vs:], Y_all[-vs:])
        fc7 = self.recursive_forecast(train_full, HORIZON)
        for i, target in enumerate(TARGETS):
            fc7[:, i] = clip_target(fc7[:, i], target)

        return {"metrics": metrics, "forecast7": fc7, "pred_int": pred_te}


# ============================================================
# 7. MODEL: VAR (многомерный, walk-forward + forecast)
# ============================================================
class VARModel:
    def __init__(self, maxlags=15):
        self.maxlags = maxlags
        self.best_lag = None

    def select_order(self, train_df):
        var_data = train_df[TARGETS].diff().dropna()
        lag_order = VAR(var_data).select_order(maxlags=self.maxlags)
        self.best_lag = lag_order.aic
        print(f"VAR best_lag (AIC) = {self.best_lag}")
        return self.best_lag

    def walk_forward(self, train_df, test_df):
        history = train_df.values.copy()
        preds, dates = [], []
        for t in range(len(test_df)):
            try:
                res  = VAR(history).fit(self.best_lag, trend='c')
                yhat = res.forecast(history[-self.best_lag:], steps=1)[0]
            except Exception:
                yhat = history[-1]
            preds.append(yhat)
            dates.append(test_df.index[t])
            history = np.vstack([history, test_df.values[t]])
        return pd.DataFrame(preds, index=dates, columns=TARGETS)

    def forecast_h(self, train_full, horizon=HORIZON):
        data = train_full[TARGETS].diff().dropna()
        res  = VAR(data).fit(self.best_lag, trend='c')
        fc   = res.forecast(data.values[-self.best_lag:], steps=horizon)
        last_vals = train_full[TARGETS].iloc[-1].values
        fc_df = pd.DataFrame(np.cumsum(fc, axis=0) + last_vals, columns=TARGETS)
        for t in TARGETS:
            fc_df[t] = clip_target(fc_df[t].values, t)
        return fc_df

    def compute_bias(self, train_full, target='meantemp', n_recent=60):
        tr  = train_full.iloc[:-n_recent]
        val = train_full[target].iloc[-n_recent:]
        data = tr[TARGETS].diff().dropna()
        res  = VAR(data).fit(self.best_lag, trend='c')
        fc   = res.forecast(data.values[-self.best_lag:], steps=n_recent)
        last_vals = tr[TARGETS].iloc[-1].values
        rec = np.cumsum(fc, axis=0) + last_vals
        return float(np.mean(val.values - rec[:, TARGETS.index(target)]))

    def run(self, train_int, test_int, train_full):
        self.select_order(train_int)

        pred_int = self.walk_forward(train_int[TARGETS], test_int[TARGETS])
        metrics = {t: evaluate(test_int[t].values, pred_int[t].values,
                               f"VAR [{t}]") for t in TARGETS}

        fc7 = self.forecast_h(train_full)
        bias = self.compute_bias(train_full, 'meantemp', n_recent=60)
        fc7_bc = fc7.copy()
        fc7_bc['meantemp'] = clip_target(fc7['meantemp'].values + bias, 'meantemp')

        return {"metrics": metrics, "forecast7": fc7, "forecast7_bc": fc7_bc,
                "bias": bias, "pred_int": pred_int}


# ============================================================
# 8. MODEL: TBATS (с bias-correction в том же блоке)
# ============================================================
class TBATSModel:
    def __init__(self, seasonal_periods=(7, 365.25)):
        self.seasonal_periods = seasonal_periods

    def _fit(self, series):
        return TBATS(seasonal_periods=list(self.seasonal_periods),
                     use_box_cox=False, use_arma_errors=False,
                     n_jobs=1).fit(series)

    def forecast_h(self, series, horizon=HORIZON):
        return self._fit(series).forecast(steps=horizon)

    def compute_bias(self, series, n_recent=60):
        tr  = series.iloc[:-n_recent]
        val = series.iloc[-n_recent:]
        preds = self._fit(tr).forecast(steps=n_recent)
        return float(np.mean(val.values - preds))

    def run(self, train_full, target='meantemp'):
        fc  = self.forecast_h(train_full[target])
        bias = self.compute_bias(train_full[target], n_recent=60)
        return {"forecast7": fc, "forecast7_bc": fc + bias, "bias": bias}


# ============================================================
# 9. MODEL: STL + SARIMAX
# ============================================================
class STLSARIMAXModel:
    def __init__(self, order=(1, 1, 1), seasonal_order=(1, 0, 1, 7), period=7):
        self.order = order
        self.seasonal_order = seasonal_order
        self.period = period

    def forecast_h(self, series, horizon=HORIZON):
        stl = STLForecast(
            series, model=ARIMA,
            model_kwargs={'order': self.order,
                          'seasonal_order': self.seasonal_order},
            period=self.period,
        )
        return stl.fit().forecast(steps=horizon)

    def run(self, train_full, target='meantemp'):
        fc = self.forecast_h(train_full[target])
        return {"forecast7": fc}


# ============================================================
# 10. RUN PIPELINE: обучить все модели и собрать метрики
# ============================================================
def run_all_models(train, test, train_int, test_int):
    results = {"sarimax": {}, "prophet": {}, "lstm": {}, "gru": {},
               "var": None, "tbats": None, "stl": None}

    print("\n===== SARIMAX =====")
    sar = SARIMAXModel()
    for t in TARGETS:
        results["sarimax"][t] = sar.run_target(t, train_int, test_int, train)
        print(f"[{t:<13s}] SARIMAX  MAPE={results['sarimax'][t]['metrics']['MAPE']:.2f}%  "
              f"bias={results['sarimax'][t]['bias']:+.2f}")

    print("\n===== Prophet =====")
    pro = ProphetModel()
    for t in TARGETS:
        results["prophet"][t] = pro.run_target(t, train_int, test_int, train)
        print(f"[{t:<13s}] Prophet  MAPE={results['prophet'][t]['metrics']['MAPE']:.2f}%  "
              f"bias={results['prophet'][t]['bias']:+.2f}")

    print("\n===== LSTM =====")
    lstm_m = MultiRNNModel('lstm').run_target_all(train_int, test_int, train)
    results["lstm"] = lstm_m
    for t in TARGETS:
        print(f"[{t:<13s}] LSTM     MAPE={lstm_m['metrics'][t]['MAPE']:.2f}%")

    print("\n===== GRU =====")
    gru_m = MultiRNNModel('gru').run_target_all(train_int, test_int, train)
    results["gru"] = gru_m
    for t in TARGETS:
        print(f"[{t:<13s}] GRU      MAPE={gru_m['metrics'][t]['MAPE']:.2f}%")

    print("\n===== VAR =====")
    var_m = VARModel().run(train_int, test_int, train)
    results["var"] = var_m
    for t in TARGETS:
        print(f"[{t:<13s}] VAR      MAPE={var_m['metrics'][t]['MAPE']:.2f}%  "
              f"bias(meantemp)={var_m['bias']:+.2f}")

    print("\n===== TBATS (meantemp) =====")
    results["tbats"] = TBATSModel().run(train, 'meantemp')
    print(f"TBATS bias: {results['tbats']['bias']:+.2f}")

    print("\n===== STL + SARIMAX (meantemp) =====")
    results["stl"] = STLSARIMAXModel().run(train, 'meantemp')

    return results


# ============================================================
# 11. ENSEMBLES + SUMMARIES
# ============================================================
def build_summary(results):
    rows = []
    for t in TARGETS:
        rows.append({**results["sarimax"][t]["metrics"], "Target": t, "Model": "SARIMAX"})
        rows.append({**results["prophet"][t]["metrics"], "Target": t, "Model": "Prophet"})
        rows.append({**results["lstm"]["metrics"][t],    "Target": t, "Model": "LSTM"})
        rows.append({**results["gru"]["metrics"][t],     "Target": t, "Model": "GRU"})
        rows.append({**results["var"]["metrics"][t],     "Target": t, "Model": "VAR"})
    df = pd.DataFrame(rows)
    pivot_mape = df.pivot_table(index="Model", columns="Target",
                                values="MAPE", aggfunc="mean").round(2)
    print("\nMAPE (%) по моделям и переменным:")
    display(pivot_mape)
    return df, pivot_mape


def build_forecast_table(results, train):
    future_dates = pd.date_range(train.index[-1] + pd.Timedelta(days=1),
                                 periods=HORIZON, freq='D')
    forecasts = {}
    for t in TARGETS:
        ti = TARGETS.index(t)
        fc = pd.DataFrame({
            "SARIMAX":  results["sarimax"][t]["forecast7"],
            "Prophet":  results["prophet"][t]["forecast7"],
            "LSTM":     results["lstm"]["forecast7"][:, ti],
            "GRU":      results["gru"]["forecast7"][:, ti],
            "VAR":      results["var"]["forecast7"][t].values,
        }, index=future_dates)

        mapes = np.array([
            results["sarimax"][t]["metrics"]["MAPE"],
            results["prophet"][t]["metrics"]["MAPE"],
            results["lstm"]["metrics"][t]["MAPE"],
            results["gru"]["metrics"][t]["MAPE"],
            results["var"]["metrics"][t]["MAPE"],
        ])
        w = 1.0 / mapes
        w = w / w.sum()
        fc["ENSEMBLE_ALL"] = (fc * w).sum(axis=1)

        # Только стат.модели
        stat_models = ["SARIMAX", "Prophet", "VAR"]
        stat_mapes  = np.array([mapes[0], mapes[1], mapes[4]])
        w_stat = (1.0 / stat_mapes) / (1.0 / stat_mapes).sum()
        fc["ENSEMBLE_STAT"] = (fc[stat_models] * w_stat).sum(axis=1)

        # Bias-corrected ансамбль (только для meantemp)
        if t == 'meantemp':
            sar_bc = results["sarimax"][t]["forecast7_bc"]
            pro_bc = results["prophet"][t]["forecast7_bc"]
            var_bc = results["var"]["forecast7_bc"][t].values
            fc["SARIMAX_BC"] = sar_bc
            fc["Prophet_BC"] = pro_bc
            fc["VAR_BC"]     = var_bc
            fc["ENSEMBLE_BC"] = (sar_bc + pro_bc + var_bc) / 3
            fc["TBATS"]      = results["tbats"]["forecast7"]
            fc["TBATS_BC"]   = results["tbats"]["forecast7_bc"]
            fc["STL+SARIMAX"] = results["stl"]["forecast7"]

        forecasts[t] = fc
    return forecasts


# ============================================================
# 12. VISUALIZATION
# ============================================================
def plot_forecasts(forecasts, train, test, target):
    fig, axes = plt.subplots(1, 2, figsize=(16, 4.5))

    # Внутренний тест
    axes[0].plot(test.index, test[target], color='black',
                 label='Факт', linewidth=1.5)
    axes[0].set_title(f"{target} — внутренний тест")
    axes[0].set_ylabel(target)

    # 7-дневный прогноз
    test_sorted = test.sort_index()
    if test_sorted.index[0] == train.index[-1]:
        actual7 = test_sorted.iloc[1:1 + HORIZON][target]
    else:
        actual7 = test_sorted.iloc[:HORIZON][target]

    axes[1].plot(train.index[-90:], train[target].iloc[-90:],
                 color='steelblue', label='История (90 дней)')
    axes[1].plot(actual7.index, actual7.values, '-o', color='black',
                 linewidth=2, label='ФАКТ')

    colors = {'SARIMAX': 'red', 'Prophet': 'green', 'LSTM': 'crimson',
              'GRU': 'darkorange', 'VAR': 'purple'}
    fc = forecasts[target]
    for model, color in colors.items():
        if model in fc.columns:
            axes[1].plot(fc.index, fc[model], '--o',
                         color=color, label=model, alpha=0.7)
    if "ENSEMBLE_STAT" in fc.columns:
        axes[1].plot(fc.index, fc["ENSEMBLE_STAT"], '-o',
                     color='navy', linewidth=2.5, label='ENSEMBLE_STAT')
    if "ENSEMBLE_BC" in fc.columns:
        axes[1].plot(fc.index, fc["ENSEMBLE_BC"], '-D',
                     color='black', linewidth=2.5, label='ENSEMBLE_BC')
    axes[1].axvline(train.index[-1], color='gray', ls=':', alpha=0.7)
    axes[1].set_title(f"{target} — прогноз на {HORIZON} дней")
    axes[1].legend(fontsize=8)
    plt.tight_layout()
    plt.show()


# ============================================================
# 13. MAIN
# ============================================================
if __name__ == "__main__":
    train, test = load_data()
    train_int, test_int = make_split(train)

    results   = run_all_models(train, test, train_int, test_int)
    summary_df, pivot_mape = build_summary(results)
    forecasts = build_forecast_table(results, train)

    for t in TARGETS:
        print(f"\n=== {t} — прогноз на {HORIZON} дней ===")
        display(forecasts[t].round(2))
        plot_forecasts(forecasts, train, test, t)
```
