```python
# Проверка магического квадрата (суммы строк, столбцов и обеих диагоналей равны)

def square(integer):
    row_sum = sum(integer[0])
    for i in integer:
        if sum(i) != row_sum:
            return False
    n = len(integer)
    for j in range(n):
        col_sum = 0
        for i in range(n):
            col_sum += integer[i][j]
        if col_sum != row_sum:
            return False
    diag1_sum = 0
    diag2_sum = 0
    for i in range(n):
        diag1_sum += integer[i][i]
        diag2_sum += integer[i][n - 1 - i]
    if diag1_sum != row_sum or diag2_sum != row_sum:
        return False

    return True

matrix = [[1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]]

square(matrix)
```

```python
# Натан, который очень любит воду (0,5 литра в час) и кататься на велосипеде

import math

def natan(hur: int) -> int:
  water = math.floor(hur / 2)
  return water

hur = int(input("Сколько часов едет Натан: "))

if 0 <= hur <= 3:
    print('Натан должен был выпить ' + str(natan(hur)) + ' литр воды')
elif 3 < hur <= 9:
    print('Натан должен был выпить ' + str(natan(hur)) + ' литра воды')
elif hur < 0 or 30 < hur:
    print('Не издевайся над беднягой Натаном!')
else:
    print('Натан должен был выпить ' + str(natan(hur)) + ' литров воды')
```

```python
# Функция, которая возвращает список целых чисел от n до 1, где n > 0

def func(n: int) -> int:
    return list(range(n, 0, -1))

e = int(input("Вставь целое число: "))
if e <= 0:
    print("Значение n не натуральное!")
else:
    print("Список целых чисел: " + str(func(e)))
```

```python
# Учитывая год, верните столетие, в котором он находится

import math

def cent(year: int) -> int:
    if year > 0:
        century = math.floor(((year - 1) / 100) + 1)
        return century
    elif year < 0:
        century = math.floor(((year - 1) / 100))
        return century

year = int(input("Укажи год: "))
if year == 0:
    print("Такого года не существует!")
else:
    print('Этот год находится в ' + str(cent(year)) + ' столетии')
```

```python
# Учитывая случайное неотрицательное число, вам необходимо вернуть цифры этого числа в массиве в обратном порядке

def str_numb(numb: int) -> list:
    if numb < 0:
        print("Число является отрицательным!")
    else:
        word = list(str(numb))
        rev = word[::-1]
        return rev

str_numb(34567)
```

```python
# Получить квадратные матрицы порядка n (через список), элементами которых являются заданные действительные числа 1, 2, ..., n*n, расположенные в ней по строкам слева направо и справа налево

def snake_hor(n):
    numbers = list(range(1, (n * n) + 1))
    matrix = []

    for i in range(n):
        row = numbers[i * n:(i + 1) * n]
        if i % 2 == 1:
            row = row[::-1]
        matrix.append(row)
    return matrix

def snake_vert(n):
    from itertools import count
    nums = count(1)
    matrix = [[0] * n for _ in range(n)]

    for col in range(n):
        if col % 2 == 0:
            for row in range(n):
                matrix[row][col] = next(nums)
        else:
            for row in range(n - 1, -1, -1):
                matrix[row][col] = next(nums)
    return matrix

n = 4
mat = snake_hor(n)
for row in mat:
    print(row)

ver = snake_vert(n)
for vert in ver:
    print(vert)
```
