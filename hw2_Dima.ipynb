```python
# Гласные vs Согласные: Посчитайте количество гласных (aoeiu) и согласных букв в строке без учёта регистра

def vowel(text):
    text = text.lower()
    vow = set('aoeiu')
    y = set('y')
    letters = [numb for numb in text if 'a' <= numb <= 'z']
    vow_count = sum(1 for numb in letters if numb in vow)
    y_count = sum(1 for numb in letters if numb in y)
    con_count = len(letters) - vow_count - y_count
    return vow_count, con_count

exam = "Happy New Year!"
vow, con = vowel(exam)
print("Количество гласных: " + str(vow) + ", " + "Количество согласных: " + str(con))
```

```python
# Удаление дублей: Удалите из строки все повторяющиеся символы, оставив только первое вхождение каждого

def non_repeat(text) -> str:
    see = set()
    res = []

    for rep in text:
        if rep not in see:
            see.add(rep)
            res.append(rep)
    return ''.join(res)

exam = "Happy New Year"
non_repeat(exam)
```

```python
# Подсчёт слов: Подсчитайте, сколько слов в строке (слова разделены пробелами, знаки препинания не учитываем)

def word_numb(text):
    space = set(' ')
    letters = [numb for numb in text if numb == ' ']
    space_count = sum(1 for numb in letters if numb in space)
    word_count = space_count + 1
    return word_count

exam = "Подсчёт слов: Подсчитайте, сколько слов в строке (слова разделены пробелами, знаки препинания не учитываем)"
"Количество слов: " + str(word_numb(exam))
```

```python
# Шифр Цезаря: Сдвиньте каждую букву строки на k позиций (например, "abc", k = 2 -> "cde")

def caesar(text: str, sft: int) -> str:
    text_up = text.upper()
    result = []
    sft = sft % 26

    for cae in text_up:
        if 'A' <= cae <= 'Z':
            start = ord('A')
            shift = (ord(cae) - start + sft) % 26 + start
            result.append(chr(shift))
        else:
            result.append(cae)
    return ''.join(result)

exam = "London is the capital of Great Britain!"
caesar(exam, 1)
```

```python
# Преобразование регистра: Сделайте заглавной первую букву каждого слова в строке (.title() запрещён)

def title(text) -> str:
    result = []
    upp = True

    for var in text:
        if var.isalpha():
            if upp:
                result.append(var.upper())
                upp = False
            else:
                result.append(var.lower())
        else:
            result.append(var)
            upp = True

    return ''.join(result)

exam = "преобразование регистра: Сделайте заглавной первую букву каждого слова в строке (.title() запрещён)"
title(exam)
```

```python
# Самая длинная подстрока: Найдите самую длинную последовательность уникальных символов подряд (например, "abcabcbb" -> "abc")

def long(text) -> str:
    start = 0
    max_len = 0
    max_start = 0
    window = set()

    for finish, numb in enumerate(text):
        while numb in window:
            window.remove(text[start])
            start += 1
        window.add(numb)
        if finish - start + 1 > max_len:
            max_len = finish - start + 1
            max_start = start
    return text[max_start:max_start + max_len]

long("asdfdsadfbbc.jkglmzkcmfdfbb")
```

```python
# Валидация скобок: Проверьте, правильно ли расставлены скобки (), [], {} в строке (используя стек-список)

def balance(text) -> bool:
    stack = []
    for par in text:
        if par in ')]}':
            stack.append(par)
        elif par == '(':
            if not stack or stack.pop() != ')':
                return False
        elif par == '[':
            if not stack or stack.pop() != ']':
                return False
        elif par == '{':
            if not stack or stack.pop() != '}':
                return False
    return not stack

balance('[]()')
```

```python
# Написать программу для решения уравнений

def square(a, b, c):

     d = b ** 2 - 4 * a * c
     if d > 0:
         x1 = (-b + d ** 0.5) / (2 * a)
         x2 = (-b - d ** 0.5) / (2 * a)
         print(f"Первый корень: {x1}, Второй корень: {x2}")
         return [x1, x2]
     elif d == 0:
         x = -b / (2 * a)
         print(f"Единственный корень: {x}")
         return [x]
     else:
         print("Нет корней!")
         return []

def repl(a, b, c):
    t_roots = square(a, b, c)

    if not t_roots:
        print("Нет корней!")
        return []

    x_roots = []
    for t in t_roots:
        x_sq = t + 8
        if x_sq < 0:
            continue
        elif x_sq == 0:
            x_roots.append(0.0)
        else:
            sqrt_val = x_sq ** 0.5
            x_roots.append(sqrt_val)
            x_roots.append(-sqrt_val)

    print("Корни: ", x_roots)
    return x_roots

square(1, -12, 20)
square(1, 17, 72)
square(1, -7, -44)
square(1, 9, 8)
square(1, -2, -63)
repl(1, 4, -5)
```

```python
# Николай задумался о поиске бесполезного числа (максимальное число, делённое на длину списка). Удачи Николаю

def useless(s: list) -> float:
    max_value = max(s)
    len_list = len(s)
    result = max_value / len_list
    return result

useless([2, 6, 7, 11, 23, 1, 5, 6, 3, 9])
```

```python
# Дополняем элементы строки разной длины, пока они не станут одинаковой длины

def all_eq(lst: list) -> list:
    max_length = max(len(text) for text in lst)
    result = []
    for var in lst:
        result.append(var + '_' * (max_length - len(var)))
    return result

all_eq(['One', 'Two', 'Three', 'Four'])
```
