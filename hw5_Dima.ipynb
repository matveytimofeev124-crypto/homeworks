```python
# Создать класс Money для работы с денежными суммами. Число должно быть представлено двумя полями: для рублей и для копеек И ВАЛЮТЫ.
# Копейки при выводе на экран должна быть отделена от целой части запятой.
# Реализовать сложение, вычитание, деление суммы на дробное число, умножение на дробное число и операции сравнения.

class Money:

    def __init__(self, rubles=0, cop=0, currency='BYN'):
        self.rubles = rubles
        self.cop = cop
        self.currency = currency
        self._normalize()

    def _normalize(self):
        total = self.rubles * 100 + self.cop
        self.rubles = total // 100
        self.cop = total % 100

    def _to_cop(self):
        return self.rubles * 100 + self.cop

    def balance(self):
        print(f'Сумма на счете {self.rubles},{self.cop:02d} {self.currency}')

    def add_money(self, rub, cop):
        self.rubles += rub
        self.cop += cop
        self._normalize()
        self.balance()
        return self.rubles, self.cop

    def substract_money(self, rub, cop):
        sub_cop = rub * 100 + cop
        if self._to_cop() < sub_cop:
            return 'Недостаточно средств'
        result_cop = self._to_cop() - sub_cop
        self.rubles = result_cop // 100
        self.cop = result_cop % 100
        self._normalize()
        self.balance()
        return self.rubles, self.cop

    def cashback_money(self, spent_rub, spent_cop, percent):
        cashback_cop = (self._to_cop()) * percent / 100
        cashback_rub = int(cashback_cop // 100)
        cashback_cop = int(cashback_cop % 100)
        print(f'Кэшбек с операции составляет {cashback_rub},{cashback_cop:02d} {self.currency}')
        self.rubles += cashback_rub
        self.cop += cashback_cop
        self._normalize()
        self.balance()
        return self.rubles, self.cop

    def multiply_money(self, percent):
        dividend_cop = (self._to_cop()) * percent / 100
        div_rub = int(dividend_cop // 100)
        div_cop = int(dividend_cop % 100)
        print(f'Начислено дивидендов: {div_rub},{div_cop:02d} {self.currency}')
        self.rubles += div_rub
        self.cop += div_cop
        self._normalize()
        self.balance()
        return self.rubles, self.cop

    def compare_money(self, other):
        diff = self._to_cop() - other._to_cop()
        if diff < 0:
            return -1
        elif diff > 0:
            return 1
        else:
            return 0

    def __eq__(self, other):
        if not isinstance(other, Money):
            return NotImplemented
        return self._to_cop() == other._to_cop()

    def __lt__(self, other):
        if not isinstance(other, Money):
            return NotImplemented
        return self._to_cop() < other._to_cop()

    def __le__(self, other):
        return self < other or self == other

    def __gt__(self, other):
        return not (self <= other)

    def __ge__(self, other):
        return not (self < other)

tom1 = Money()
tom1.add_money(2, 230)
tom1.substract_money(1, 50)
tom1.cashback_money(100, 0, 3)
tom1.multiply_money(5)

tom2 = Money(2, 80)
print(tom1.compare_money(tom2))
print(tom1 == tom2)
print(tom1 > Money(2, 50))
```
