```python
import math # Задача 1
x = 1.79
var_1 = math.cos(math.exp(x))
var_2 = math.log((1 + x)**2)
var_3 = math.sqrt(math.exp(math.cos(x)) + ((1 - math.cos(2 * math.pi * x)) / 2))
var_4 = math.sqrt(1 / x) + math.cos(x**2)
var_5 = (var_1 + var_2 + var_3 + var_4) ** math.sin(x)
print(var_5)
```
