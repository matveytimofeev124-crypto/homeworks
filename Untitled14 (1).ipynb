```python
# !pip -q install datasets transformers evaluate seqeval accelerate

import os, json, re, time, random, gc
import numpy as np
import torch
from collections import Counter
from datasets import load_dataset
from transformers import (
    AutoTokenizer, AutoModelForTokenClassification,
    TrainingArguments, Trainer, DataCollatorForTokenClassification,
)
from evaluate import load as load_metric

# --- HF_TOKEN (раскомментируйте, если есть) ---
# from huggingface_hub import login
# login(token="hf_xxxxxxxxxxxxxxxxxxxx")

def set_seed(seed=42):
    random.seed(seed); np.random.seed(seed)
    torch.manual_seed(seed); torch.cuda.manual_seed_all(seed)

set_seed(42)
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("CUDA:", torch.cuda.is_available(),
      "|", torch.cuda.get_device_name(0) if torch.cuda.is_available() else "CPU")
```

```python
# 2.1. Загрузка
raw_train = load_dataset("redmadrobot-rnd/pii_train", split="train")
raw_bench = load_dataset("redmadrobot-rnd/pii_benchmark", split="test")

def parse_json_fields(ex):
    ex["tokens"]   = json.loads(ex["tokens"])
    ex["ner_tags"] = json.loads(ex["ner_tags"])
    return ex

train_ds = raw_train.map(parse_json_fields)
bench_ds = raw_bench.map(parse_json_fields)

# 2.2. Tag-aware IP-merge: склеиваем IPv4 ТОЛЬКО внутри B-IP_ADDRESS
def merge_ip_tokens(tokens, tags):
    out_t, out_g, i = [], [], 0
    while i < len(tokens):
        is_ip_pattern = (i + 6 < len(tokens)
            and tokens[i].isdigit() and tokens[i+1] == "." and tokens[i+2].isdigit()
            and tokens[i+3] == "." and tokens[i+4].isdigit() and tokens[i+5] == "."
            and tokens[i+6].isdigit())
        if is_ip_pattern and tags[i] == "B-IP_ADDRESS":
            out_t.append("".join(tokens[i:i+7])); out_g.append(tags[i]); i += 7
        else:
            out_t.append(tokens[i]); out_g.append(tags[i]); i += 1
    return out_t, out_g

def merge_ip_in_dataset(examples):
    new_tokens, new_tags = [], []
    for toks, tags in zip(examples["tokens"], examples["ner_tags"]):
        t, g = merge_ip_tokens(toks, tags)
        new_tokens.append(t); new_tags.append(g)
    return {"tokens": new_tokens, "ner_tags": new_tags}

train_ds = train_ds.map(merge_ip_in_dataset, batched=True)
bench_ds = bench_ds.map(merge_ip_in_dataset, batched=True)

# 2.3. Метки
all_tags = set()
for ex in train_ds: all_tags.update(ex["ner_tags"])
for ex in bench_ds: all_tags.update(ex["ner_tags"])
label_list = sorted(all_tags)
label2id = {l: i for i, l in enumerate(label_list)}
id2label = {i: l for i, l in enumerate(label_list)}
print(f"Меток: {len(label_list)} → {len(label_list)//2 - 1} типов + 'O'")

# 2.4. Split
split = train_ds.train_test_split(test_size=0.1, seed=42)
train_data, val_data = split["train"], split["test"]
print(f"Train: {len(train_data)}, Val: {len(val_data)}, Bench: {len(bench_ds)}")

# 2.5. Частоты классов
type_freq = Counter()
for ex in train_data:
    for tag in ex["ner_tags"]:
        if tag != "O":
            type_freq[tag[2:]] += 1
print(f"Типов: {len(type_freq)}")
```

```python
def tokenize_data(examples, tokenizer, with_labels=True):
    tok = tokenizer(
        examples["tokens"], truncation=True, max_length=512,
        is_split_into_words=True,
    )
    word_ids_all, labels_all = [], []
    for i, seq in enumerate(examples["ner_tags"]):
        wids = tok.word_ids(batch_index=i)
        word_ids_all.append(wids)
        if with_labels:
            seq_ids = [label2id[t] for t in seq]
            prev, aligned = None, []
            for w in wids:
                if w is None:   aligned.append(-100)
                elif w != prev: aligned.append(seq_ids[w])
                else:           aligned.append(-100)
                prev = w
            labels_all.append(aligned)
    tok["word_ids"] = word_ids_all
    if with_labels:
        tok["labels"] = labels_all
    tok.pop("tokens", None); tok.pop("ner_tags", None)
    return tok
```

```python
METRIC = load_metric("seqeval")

def compute_metrics(eval_pred):
    preds, labels = eval_pred
    preds = np.argmax(preds, axis=2)
    tp = [[label_list[p] for p,l in zip(pr,la) if l != -100] for pr,la in zip(preds,labels)]
    tl = [[label_list[l] for p,l in zip(pr,la) if l != -100] for pr,la in zip(preds,labels)]
    r = METRIC.compute(predictions=tp, references=tl)
    return {"precision": r["overall_precision"], "recall": r["overall_recall"],
            "f1": r["overall_f1"], "accuracy": r["overall_accuracy"]}


def train_ner_model(checkpoint, output_dir, num_epochs=3, lr=2e-5, batch_size=None, seed=42):
    print(f"\n{'='*60}\nОбучаем: {checkpoint}\n{'='*60}")
    set_seed(seed); t0 = time.time()

    tok = AutoTokenizer.from_pretrained(checkpoint)
    train_tok = train_data.map(lambda x: tokenize_data(x, tok), batched=True, remove_columns=["text"])
    val_tok   = val_data.map(  lambda x: tokenize_data(x, tok), batched=True, remove_columns=["text"])

    model = AutoModelForTokenClassification.from_pretrained(
        checkpoint, num_labels=len(label_list), id2label=id2label, label2id=label2id,
    )
    bs = batch_size or (16 if DEVICE == "cuda" else 8)

    args = TrainingArguments(
        output_dir=output_dir, learning_rate=lr,
        per_device_train_batch_size=bs, per_device_eval_batch_size=bs,
        num_train_epochs=num_epochs, weight_decay=0.01,
        eval_strategy="epoch", save_strategy="epoch",
        load_best_model_at_end=True, metric_for_best_model="f1",
        save_total_limit=1,
        logging_steps=100, report_to="none",
        fp16=(DEVICE == "cuda"), seed=seed,
    )
    trainer = Trainer(
        model=model, args=args,
        train_dataset=train_tok, eval_dataset=val_tok,
        data_collator=DataCollatorForTokenClassification(tok),
        compute_metrics=compute_metrics,
    )
    trainer.train()
    trainer.save_model(output_dir); tok.save_pretrained(output_dir)
    val_metrics = trainer.evaluate()

    # Освобождаем GPU
    trainer.model.cpu()
    del model, trainer, train_tok, val_tok
    gc.collect(); torch.cuda.empty_cache()
    free, _ = torch.cuda.mem_get_info() if DEVICE == "cuda" else (0, 0)
    print(f"⏱ {(time.time()-t0)/60:.1f} мин | Val F1: {val_metrics['eval_f1']:.4f} | "
          f"GPU free: {free/1e9:.2f} GB")
    return output_dir
```

```python
def evaluate_on_benchmark(model_dir, bench_ds, apply_url_filter=True, batch_size=32, verbose=False):
    tok = AutoTokenizer.from_pretrained(model_dir)
    model = AutoModelForTokenClassification.from_pretrained(model_dir).to(DEVICE).eval()

    tok_bench = bench_ds.map(lambda x: tokenize_data(x, tok, with_labels=False), batched=True)
    word_ids_all = tok_bench["word_ids"]
    keep = [c for c in tok.model_input_names if c in tok_bench.column_names]
    tok_bench = tok_bench.remove_columns([c for c in tok_bench.column_names if c not in keep])

    def pad_batch(batch_dict):
        max_len = max(len(x) for x in batch_dict["input_ids"])
        return {k: torch.tensor(
            [x + [tok.pad_token_id if k == "input_ids" else 0] * (max_len - len(x))
             for x in batch_dict[k]], dtype=torch.long).to(DEVICE) for k in keep}

    all_pred_ids = []
    with torch.no_grad():
        for start in range(0, len(tok_bench), batch_size):
            batch = {k: tok_bench[k][start:start+batch_size] for k in keep}
            all_pred_ids.extend(model(**pad_batch(batch)).logits.argmax(dim=-1).cpu().tolist())

    def unalign(pred_row, wids, n):
        out, prev = [None]*n, None
        for pos, w in enumerate(wids):
            if w is None: continue
            if w != prev: out[w] = id2label[pred_row[pos]]
            prev = w
        return [t if t else "O" for t in out]

    def filter_url(tags, tokens):
        """Убирает B-URL-спаны, внутри которых нет точки/двоеточия/w."""
        out = tags[:]; i = 0
        while i < len(out):
            if out[i] == "B-URL":
                j = i + 1
                while j < len(out) and out[j] == "I-URL": j += 1
                if not any(c in " ".join(tokens[i:j]) for c in (".", ":", "w", "W")):
                    for k in range(i, j): out[k] = "O"
                i = j
            else:
                i += 1
        return out

    all_pred, all_true = [], []
    for i, ex in enumerate(bench_ds):
        p = unalign(all_pred_ids[i], word_ids_all[i], len(ex["tokens"]))
        if apply_url_filter:
            p = filter_url(p, ex["tokens"])
        all_pred.append(p); all_true.append(ex["ner_tags"])

    r = METRIC.compute(predictions=all_pred, references=all_true)

    if verbose:
        print(f"\n{'Тип':<20} {'P':>6} {'R':>6} {'F1':>6} {'N':>5}")
        print("-" * 46)
        for et in sorted(type_freq):
            if et in r:
                m = r[et]
                print(f"{et:<20} {m['precision']:>6.3f} {m['recall']:>6.3f} {m['f1']:>6.3f} {m['number']:>5}")

    del model, tok_bench; gc.collect(); torch.cuda.empty_cache()
    return r
```

```python
CONFIGS = [
    ("RuBERT",       "DeepPavlov/rubert-base-cased",              "./rubert-pii-ner",     2e-5, None),
    ("DistilBERT",   "Geotrend/distilbert-base-ru-cased",         "./distilbert-pii-ner", 3e-5, None),
    ("mBERT",        "google-bert/bert-base-multilingual-cased",  "./mbert-pii-ner",      2e-5, None),
    ("XLM-R base",   "FacebookAI/xlm-roberta-base",               "./xlmr-pii-ner",       2e-5, None),
    # ("XLM-R large", "FacebookAI/xlm-roberta-large", "./xlmr-large-pii-ner", 1e-5, 8),
]

model_dirs = {}
for name, ckpt, out, lr, bs in CONFIGS:
    model_dirs[name] = train_ner_model(ckpt, out, num_epochs=3, lr=lr, batch_size=bs, seed=42)
print("\nОбучены:", list(model_dirs))
```

```python
results_raw, results_filtered = {}, {}
for name, mdir in model_dirs.items():
    print(f"\n{'#'*60}\n# {name}\n{'#'*60}")
    results_raw[name]      = evaluate_on_benchmark(mdir, bench_ds, apply_url_filter=False)
    results_filtered[name] = evaluate_on_benchmark(mdir, bench_ds, apply_url_filter=True)

print("\n" + "="*82)
print(f"{'Модель':<15} {'Raw F1':>10} {'+URLf F1':>10} {'Δ filter':>10} | "
      f"{'+URLf P':>10} {'+URLf R':>10}")
print("-"*82)
for name in results_raw:
    a, b = results_raw[name], results_filtered[name]
    print(f"{name:<15} {a['overall_f1']:>10.4f} {b['overall_f1']:>10.4f} "
          f"{b['overall_f1']-a['overall_f1']:>+10.4f} | "
          f"{b['overall_precision']:>10.4f} {b['overall_recall']:>10.4f}")

# Per-class для двух лучших
for name in ("XLM-R base", "RuBERT"):
    print(f"\n### {name} — per class")
    evaluate_on_benchmark(model_dirs[name], bench_ds, apply_url_filter=True, verbose=True)
```

```python
# ---------- Утилиты предобработки ----------
SNILS_RE = re.compile(r"^(\d{3})-(\d{3})-(\d{3})([.,;:!?]*)$")
PUNCT_TRAIL = ".,;:!?)]}\"'"

def normalize_text(text):
    """Склеиваем разнесённые пробелами IP-адреса."""
    return re.sub(r"\b(\d+)\s*\.\s*(\d+)\s*\.\s*(\d+)\s*\.\s*(\d+)\b",
                  r"\1.\2.\3.\4", text)

def smart_pretokenize(text):
    """
    Приводит raw-текст к тренировочному формату:
    - IP склеивает в 1 токен (192.168.1.1)
    - SNILS разбивает на 6 токенов (123 - 456 - 789 00)
    - Отделяет хвостовую пунктуацию (00, → 00 + ,)
    """
    raw = text.split()
    out, i = [], 0
    while i < len(raw):
        tok = raw[i]

        # 1) IP-merge
        if (i + 6 < len(raw) and raw[i].isdigit() and raw[i+1] == "."
            and raw[i+2].isdigit() and raw[i+3] == "."
            and raw[i+4].isdigit() and raw[i+5] == "." and raw[i+6].isdigit()):
            out.append("".join(raw[i:i+7])); i += 7; continue

        # 2) SNILS-split
        m = SNILS_RE.match(tok)
        if m:
            out.extend([m.group(1), "-", m.group(2), "-", m.group(3)])
            if m.group(4):
                out.append(m.group(4))
            i += 1; continue

        # 3) Хвостовая пунктуация отдельно
        m = re.match(r"^(.*?)([.,;:!?]+)$", tok)
        if m and m.group(1):
            out.append(m.group(1)); out.append(m.group(2))
        else:
            out.append(tok)
        i += 1
    return out


# ---------- Span-логика ----------
def tokens_to_spans(text, tokens, tags):
    spans, cur, offset = [], None, 0
    for tok, tag in zip(tokens, tags):
        start = text.find(tok, offset)
        if start == -1: start = offset
        end = start + len(tok); offset = end
        if tag.startswith("B-"):
            if cur: spans.append(cur)
            cur = {"type": tag[2:], "start": start, "end": end, "text": tok}
        elif tag.startswith("I-") and cur and cur["type"] == tag[2:]:
            cur["end"] = end; cur["text"] = text[cur["start"]:end]
        else:
            if cur: spans.append(cur); cur = None
    if cur: spans.append(cur)
    return spans


def replace_pii(text, entities, mapping=None):
    """Замена спанов на теги с сохранением пунктуации."""
    if mapping is None:
        mapping = {t: f"[{t}]" for t in type_freq}
    for e in sorted(entities, key=lambda x: x["start"], reverse=True):
        tag = mapping.get(e["type"], f"[{e['type']}]")
        frag = text[e["start"]:e["end"]]
        core = frag.rstrip(PUNCT_TRAIL); trail = frag[len(core):]
        lead = core[:len(core) - len(core.lstrip("([{'\""))]
        text = text[:e["start"]] + lead + tag + trail + text[e["end"]:]
    return text


# ---------- Sliding window inference ----------
def predict_long_tokens(tokens, model_dir, max_words=500, stride_words=100):
    """
    Sliding window для длинных последовательностей.
    Возвращает BIO-теги для каждого слова через majority vote по перекрытиям.
    """
    tok = AutoTokenizer.from_pretrained(model_dir)
    model = AutoModelForTokenClassification.from_pretrained(model_dir).to(DEVICE).eval()

    def _predict_window(w_tokens):
        enc = tok(w_tokens, is_split_into_words=True, truncation=True,
                  max_length=512, return_tensors="pt")
        wids = enc.word_ids(0)
        with torch.no_grad():
            inputs = {k: v.to(DEVICE) for k, v in enc.items()}
            preds = model(**inputs).logits.argmax(dim=-1)[0].cpu().tolist()
        out, prev = [None] * len(w_tokens), None
        for pos, w in enumerate(wids):
            if w is None: continue
            if w != prev: out[w] = id2label[preds[pos]]
            prev = w
        return [t if t else "O" for t in out]

    # Короткий текст — одно окно
    if len(tokens) <= max_words:
        result = _predict_window(tokens)
        del model, tok; gc.collect(); torch.cuda.empty_cache()
        return result

    # Длинный — окна + majority vote
    votes = [Counter() for _ in range(len(tokens))]
    step = max_words - stride_words
    for start in range(0, len(tokens), step):
        end = min(start + max_words, len(tokens))
        window_tags = _predict_window(tokens[start:end])
        for local_idx, tag in enumerate(window_tags):
            votes[start + local_idx][tag] += 1

    final = []
    for c in votes:
        if not c:
            final.append("O"); continue
        best = max(c.items(), key=lambda kv: (
            kv[1], kv[0] != "O", kv[0].startswith("B-")
        ))
        final.append(best[0])

    del model, tok; gc.collect(); torch.cuda.empty_cache()
    return final


# ---------- End-to-end ----------
def tag_text(text, model_dir, max_words=500, stride_words=100):
    text = normalize_text(text)
    tokens = smart_pretokenize(text)
    tags = predict_long_tokens(tokens, model_dir, max_words, stride_words)
    spans = tokens_to_spans(text, tokens, tags)
    return replace_pii(text, spans), spans


# ---------- Демо на XLM-R ----------
MODEL = model_dirs["XLM-R base"]
examples = [
    "Мой ИНН 7707083893, СНИЛС 123-456-789 00, карта 4276 1234 5678 9010.",
    "Звоните Петровой Марии по телефону +7 (495) 123-45-67, она живёт на ул. Ленина, д. 5.",
    "Сервер доступен по адресу 192.168.1.1, логи на https://example.com/logs.",
    "Отправьте результат Ивану на ivan@example.com",
]
for txt in examples:
    masked, _ = tag_text(txt, MODEL)
    print("БЫЛО:  ", txt)
    print("СТАЛО:", masked)
    print("-" * 70)

# ---------- Sliding window на длинном документе ----------
print("\n" + "="*70)
print("SLIDING WINDOW: документ из 5 примеров бенчмарка")
print("="*70)
long_text = "\n\n".join(bench_ds[i]["text"] for i in range(5))
tok_probe = AutoTokenizer.from_pretrained(MODEL)
n_tokens = len(tok_probe(long_text.split(), is_split_into_words=True)["input_ids"])
print(f"Символов: {len(long_text)}, слов: {len(long_text.split())}, токенов: {n_tokens}")
if n_tokens > 512:
    print(f"⚠️  Без sliding window {n_tokens - 512} токенов в хвосте потерялись бы")

masked_long, spans_long = tag_text(long_text, MODEL)
print(f"Найдено сущностей: {len(spans_long)}")
print(f"По типам: {dict(Counter(s['type'] for s in spans_long))}")
print(f"\nХвост маскированного документа:\n...{masked_long[-800:]}")
```
