```python
import numpy as np
```

```python
A = np.array([[1, 2, 4],
              [7, 5, -2],
              [-2, 1, 3]])

B = np.array([[-1, 0, 3],
              [2, -6, 1],
              [-7, 1, 5]])

C = np.array([[-2, 3],
              [4, 0],
              [7, -1]])

result1 = A + B
result2 = B - A
result3 = A @ C
result4 = A @ B @ C
print(f"Результат сложения матриц A и B: \n{result1}")
print(f"Результат вычитания матриц B и A: \n{result2}")
print(f"Результат умножения матриц A и C: \n{result3}")
print(f"Результат умножения матриц A, B и C: \n{result4}")
```

```python
A = np.array([[4, 0, 4, 5],
              [0, 1, 5, 6],
              [-2, 2, 2, 0],
              [1, -5, 1, 3]])

A_inv = np.linalg.inv(A)
det = np.linalg.det(A)
print(f"Определитель: {det:.2f}")
print("Обратная матрица A:")
print(np.round(A_inv, 2))
I = A @ A_inv
print("Проверка единичной матрицы:")
print(np.round(I, 0))
```
